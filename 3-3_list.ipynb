## 4. 리스트 관련 함수

리스트에 요소 추가 : append(value)

```python
a = [1,2,3]

a.append("four")
a
```

```python
a.append([5,6])
a
```

 리스트 정렬 : sort()

```python
a = [2,4,9,5,6,7]
a.sort()
a
```

```python
a = ['e','y','a','q']
a.sort()
a
```

```python
a.sort(reverse=True)
a
```

 리스트 뒤집기 (인덱스 역순 정렬) : reverse ()

```python
a = ['e','y','a','q']
a.reverse()
a
```

 인덱스 반환 : index (value)

```python
a
```

```python
a.index('q')
```

```python
a.index('b')
```

 리스트에 요소 삽입 : index (index, value)

```python
a = [1,2,3]

a.insert(0,4)
a
```

```python
a.insert(3,5)
a
```

 리스트 요소 제거 : remove (value)

```python
a = ['e','y','a','q']

a.remove('a')
a
```

```python
a = ['e','y','a','q','y','a','q']
a.remove('q')
a
```

 리스트 요소 끄집어 내기 : pop (index)

```python
a = [1,2,3]

a.pop()
```

```python
a
```

```python
a = [1,2,3]

a.pop(0)
```

```python
a
```

 리스트에 포함된 요소의 개수 세기 : count (value)

```python
a = [1,2,3,1,1,1,1]
a.count(1)
```

 리스트 확장 : extend (value)

```python
a = [1,2,3]
a.extend([4,5])
a
```

```python
b = [6,7]
a.extend(b)
a
```

```python
a = [1,2,3]
a += [4,5] # a = a +| [4,5]
a
```
