datetime.date

ex) A 군과 B 양이 2024년 12월 14일부터 만나기 시작했다면 2026년 4월 5일은 둘이 사귄 지 며칠째 되는 날일까? 아울러 사귀기 시작한 2024년 12월 14일은 무슨 요일이었을까? datetime.date 함수를 사용하면 이 문제를 쉽게 해결할 수 있다.

```python
import datetime

day1 = datetime.date(2024,12,14)
day2 = datetime.date(2026,4,5)

diff = day2 -day1
diff
```

### random : 난수(규칙이 없는 임의의 수)를 발생

```python
import random

random.random()  # 0 <= x < 1.0
```

```python
# randint () : 정수 중에서 난수 발생
random.randint(1,10)  # 1 <= x <= 10
```

```python
# random_pop 함수
def random_pop(data) : 
    num = random.randint(0,len(data)-1)
    print(data, num)
    return data.pop(num)

data = [1,2,3,4,5]
while data:
    print(random_pop(data))
```

random.choice() : 입력받은 리스트에서 무작위로 하나의 요소를 선택해 리턴

```python
def random_pop(data) : 
    num = random.choice(data)
    print(data, num)
    data.remove(num)
    return num

data = [1,2,3,4,5]
while data:
    print(random_pop(data))
```

random.sample() : 리스트 요소를 무작위로 석어서 반환

```python
data = [1,2,3,4,5]
random.sample(data, len(data))
```

json : JSON 데이터를 처리할 때 사용

```python
import json

with open('myinfo.json', 'r', encoding='utf-8') as f:
    data = json.load(f)
```

```python
data
```

```python
type(data)
```

```python
# dict 자료형 -> json 파일로
data = {'name': '홍길동', 'birth':'0525', 'age':20}

with open ('myinfo.json','w') as f:
    json.dump(data, f)
```

```python
d = {'name': '홍길동', 'birth':'0525', 'age':20}
json_data=json.dumps(d)

json_data
```

```python
# json => dict
json.loads(json_data)
```

```python
d = {'name': '홍길동', 'birth':'0525', 'age':20}
json_data=json.dumps(d, ensure_ascii=False)

json_data
```

```python
json.loads(json_data)
```

```python
d = {'name': '홍길동', 'birth':'0525', 'age':20}
json_data=json.dumps(d, ensure_ascii=False, indent=2)

print(json_data)
```

zipfile : 여러개 파일을 zip 형식으로 합치거나 해제

```python
import zipfile

with zipfile.ZipFile('mytest.zip','w') as myzip:
    myzip.write('a.txt')
    myzip.write('b.txt')
    myzip.write('c.txt')
```

```python
#압축 풀기
with zipfile.ZipFile('mytest.zip') as myzip:
    myzip.extractall()
```
