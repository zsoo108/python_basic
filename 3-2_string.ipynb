## 3. 여러 줄의 문자열을 변수에 대입하고 싶을 때

```python
# 이스케이프 코드 \n 삽입
multiline = "life is too short\nYou need python"
print(multiline)
```

```python
# 연속된 따옴표 3개 사용
multiline = """
life is too short
You need python."""
print(multiline)
```

```python
multiline = '''
life is too short
You need python.'''
print(multiline)
```

## 7. 문자열 포매팅

### 1) 숫자 바로 대입

```python
"i eat %d apples." %3
```

```python
"i eat %d apples." %5
```

### 2) 문자열 바로 대입

```python
"i eat %d apples." %five
```

```python
"i eat %s apples." %"five"
```

### 3) 숫자 값을 나타내는 변수로 대입

```python
num = 3
"i eat %d apples." %num
```

### 4) 2개 이상의 값 넣기

```python
number = 10
day = 'three'
"I ate %d apples, so i was sick for %s days."%(number, day)
```

```python
"I ate %d apples, so i was sick for %s days."%(day, number)
```

## 10. format() 함수를 사용한 포매팅

```python
"i eat {} apples.".format(3)
```

```python
"i eat {} apples.".format('five')
```

```python
num = 3
"i eat {} apples.".format(num)
```

```python
number = 10
day = 'three'
"I ate {0} apples, so i was sick for {1} days.".format(number, day)
```

```python
# 변수명으로 넣기
"I ate {number} apples, so i was sick for {day} days.".format(day=3, number=10)
```

```python
# 인덱스와 변수명을 혼용해서 넣기
"I ate {0} apples, so i was sick for {day} days.".format(10, day=3)
```

```python
# 왼쪽 정렬
"{:<10}".format('hi')
```

```python
# 오른쪽 정렬
"{:>10}".format('hi')
```

```python
# 가운데 정렬
"{:^10}".format('hi')
```

```python
# 공백 채우기
"{:=^10}".format('hi')
```

```python
"{:!<10}".format('hi')
```

```python
# 소수점 표현
"{:0.4f}".format(2.3414)
```

```python
"{:10.4f}".format(3.223131)
```

```python
# {또는} 문자 표현
"{{and}}".format()
```

## 12. 문자열 관련 함수들

### count () : 특정 문자 개수 세기

```python
a = "hobby"
a.count('b')
```

### find () : 특정 문자 위치 찾기

```python
a = "Python is the best choice"
a.find('b')
```

```python
a.find('k')
```

### index () : 특정 문자 위치 찾기

```python
a.index('b')
```

```python
a.index('k')
```

### join () : 문자열 구분자 삽입하여 연결

```python
", ".join('abcd')
```

```python
", ".join(['a','b','c','d'])
```

### upper () : 대문자로

```python
"hi".upper()
```

### lower () : 소문자로

```python
"HI".lower()
```

### lstrip (), rstrip (), strip () : 공백 제거

```python
a = " hi "
a.lstrip()
```

```python
a = " hi "
a.rstrip()
```

```python
a = " hi "
a.strip()
```

### replace (): 문자열 바꾸기

```python
a = "Life is too short"
b = a.replace("Life", "your leg")
b
```

### split () : 문자열 나누기

```python
# 공백 기준으로 나누기
a.split()
```

```python
# 구분자로 나누기
"a:b:c:d".split(':')
```

```python
"a,b,c,d".split(',')
```
