## 5. 딕셔너리 관련 함수

 key 리스트 만들기 : keys()

```python
a = {'name': 'hjs', 'phone':'010-4622-2639', 'birth':'1008'}
```

```python
a.keys()
```

```python
type(a.keys())
```

```python
list(a.keys())
```

```python
type(list(a.keys()))
```

```python
for k in a.keys():
    print(k)
```

 value 리스트 만들기 : values()

```python
a.values()
```

key, value 쌍 얻기 : items()

```python
a.items()
```

key, value 쌍 모두 지우기:clear()

```python
a.clear()
a
```
